```python
import numpy as np
import matplotlib.pyplot as plt
customer_genre = np.array([
    [5, 3, 0, 1, 2],
    [4, 0, 2, 1, 0],
    [1, 5, 4, 0, 2],
    [0, 2, 5, 4, 1],
    [3, 1, 0, 5, 4],
    [2, 4, 1, 3, 5]
], dtype=float)
genre_book = np.array([
    [0.9, 0.2, 0.1, 0.3],
    [0.1, 0.8, 0.2, 0.2],
    [0.2, 0.1, 0.9, 0.2],
    [0.1, 0.2, 0.1, 0.9],
    [0.3, 0.2, 0.2, 0.4]
])
print("Customer-Genre Matrix:\n", customer_genre)
print("\nGenre-Book Matrix:\n", genre_book)
print("\nDot Product:",
      np.dot(customer_genre[0], customer_genre[1]))
score = np.dot(customer_genre, genre_book)
print("\nCustomer-Book Score Matrix:\n", score)
print("\nTranspose:\n", customer_genre.T)
A = np.array([[4, 2], [1, 3]])
print("\nInverse:\n", np.linalg.inv(A))
print("\nAddition:\n", customer_genre + 2)
print("\nMultiplication:\n", customer_genre * 2)
norm = np.linalg.norm(customer_genre, axis=1, keepdims=True)
normalized = customer_genre / np.maximum(norm, 1e-12)
similarity = normalized @ normalized.T
print("\nCosine Similarity:\n", similarity)
i, j = np.unravel_index(
    np.argmax(np.triu(similarity, k=1)),
    similarity.shape
)
print("Most Similar Customers:", i+1, "and", j+1)
x = customer_genre[0].reshape(-1, 1)
W = np.array([
    [0.2, 0.1, 0.3, 0.4, 0.2],
    [0.1, 0.3, 0.2, 0.1, 0.4],
    [0.3, 0.2, 0.1, 0.2, 0.1]
])
b = np.array([[0.1], [0.2], [0.3]])
y = W @ x + b
print("\nDense Layer Output:\n", y)
plt.imshow(score, cmap="viridis", aspect="auto")
plt.colorbar()
plt.xlabel("Books")
plt.ylabel("Customers")
plt.title("Customer-Book Recommendation Scores")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

# Generate 200 patient records
n = 200

data = pd.DataFrame({
    "Haemoglobin": np.random.normal(13, 1.5, n),
    "RBC": np.random.normal(4.5, 0.5, n),
    "Glucose": np.random.normal(100, 15, n),
    "Cholesterol": np.random.normal(190, 25, n),
    "BMI": np.random.normal(25, 4, n)
})

print(data.head())

# Standardisation
scaler = StandardScaler()
X = scaler.fit_transform(data)

# Covariance
covariance = np.cov(X, rowvar=False)

# Correlation
correlation = np.corrcoef(X, rowvar=False)

print("\nCovariance Matrix:\n", covariance)
print("\nCorrelation Matrix:\n", correlation)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.heatmap(covariance, annot=True, cmap="coolwarm")
plt.title("Covariance Matrix")
plt.subplot(1, 2, 2)
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
eigenvalues, eigenvectors = np.linalg.eigh(covariance)
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
variance = eigenvalues / np.sum(eigenvalues) * 100
print("\nVariance Percentage:")
for i, v in enumerate(variance):
    print(f"PC{i+1}: {v:.2f}%")
projected = X @ eigenvectors[:, :2]
plt.scatter(projected[:, 0], projected[:, 1])
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Patient Data")
plt.grid()
plt.show()
plt.figure(figsize=(7, 5))
plt.axhline(0, color="gray")
plt.axvline(0, color="gray")

for i in range(2):
    plt.arrow(0, 0,
              eigenvectors[0, i],
              eigenvectors[1, i],
              head_width=0.05)
    plt.text(eigenvectors[0, i],
             eigenvectors[1, i],
             f"PC{i+1}")

plt.title("First Two Eigenvectors")
plt.xlabel("Haemoglobin Loading")
plt.ylabel("RBC Loading")
plt.grid()
plt.show()
```
